```python
horas = int(input("Digite o tanto de horas: "))

if horas <= 2:
    valor = 8.00
elif horas <= 5 and horas > 2:
    valor = 8 +(horas - 2) * 3
else:
    valor = 17 + (horas - 5) * 5

print("O valor a ser pago é:", int(valor))
#o calculo não é igual para todas as horas
#existem 3 faixas: abaixo de dois horas; de dois a cinco; e acima de cinco.
```

```python
anterior = int(input("Número: "))
repetido = False

for _ in range(7):
    atual = int(input("Número: "))
    if atual == anterior:
        repetido = True
    anterior = atual

if repetido:
    print("Existem valores consecutivos iguais")
else:
    print("Não existem valores consecutivos iguais")
```

```python
saldo = 500.0

for _ in range(6):
    tipo = int(input("1=Depósito, 2=Saque: "))
    valor = float(input("Valor: "))

    if tipo == 1:
        saldo += valor
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente")

print("Saldo final: ", saldo)
#o saldo é o estado atual, pois o valor pode mudar conforme as operações são feitas
#quando o saque não puder ser realizado informa que não se tem o saldo suficiente
```

```python
num = int(input("Digitei um número maior que 1: "))
primo = True

for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
        primo = False
        break

print("É primo" if primo else "Não é primo")
```

```python
n = int(input("Digite um número positivo:"))
quantidade = 0
print("Divisores:")
for i in range(1, n + 1):
    if n % i == 0:
        print(i, end=" ")
        quantidade += 1
print("Total de divisores:", quantidade)
```
