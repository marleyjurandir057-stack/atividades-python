```python
numeros = []
pares = []
impares = []
for i in range(10):
    num = int(input("Digite um número inteiro: "))
    numeros.append(num)
for num in numeros:
    if num % 2 == 0:
        pares.append(num)
    else:
        impares.append(num)
print("Todos os números:", numeros)
print("Números pares:", pares)
print("Números ímpares:", impares)
```

```python

lista_temp = []
for i in range(6):
    num = int(input("Digite um número inteiro: "))
    lista_temp.append(num)

numeros = tuple(lista_temp)
qtd_cinco = 0
qtd_pares = 0
pos_cinco = -1
for i in range(len(numeros)):
    if numeros[i] == 5:
        qtd_cinco += 1
        if pos_cinco == -1:
            pos_cinco = i
    if numeros[i] % 2 == 0:
        qtd_pares += 1
print("Tupla:", numeros)
print("Quantidade de vezes que o 5 apareceu:", qtd_cinco)
if pos_cinco != -1:
    print("Posição da primeira ocorrência do 5:", pos_cinco)
else:
    print("O número 5 não foi digitado.")

print("Quantidade de números pares:", qtd_pares)
```

```python
contagem = {}

print("Digite 8 palavras:")
for i in range(8):
    palavra = input("Palavra " + str(i + 1) + ": ").lower()
    if palavra in contagem:
        contagem[palavra] += 1
    else:
        contagem[palavra] = 1
print("Frequência das palavras:")
for palavra, qtd in contagem.items():
    print(palavra + ": " + str(qtd) + " vezes")
```

```python
estoque = {
    "caneta": 50,
    "caderno": 20,
    "borracha": 30,
    "lápis": 40
}
produto = input("Nome do produto vendido: ").lower()
qtd_vendida = int(input("Quantidade vendida: "))

if produto in estoque:
    if estoque[produto] >= qtd_vendida:
        estoque[produto] -= qtd_vendida
        print("Venda realizada com sucesso")
    else:
        print("Estoque insuficiente")
else:
    print("Produto não cadastrado")

print("Estoque final: ", estoque)
```

```python
catalogo = []
print("Cadastro de 5 Produto")
for i in range(5):
    codigo = int(input("Código: "))
    nome = input("Nome: ")
    preco = float(input("Preço: "))
    prod_tupla = (codigo, nome, preco)
    catalogo.append(prod_tupla)
busca_codigo = int(input("\nDigite o código do produto para buscar: "))
achou = False
for prod in catalogo:
    if prod[0] == busca_codigo:
        print("Produto encontrado:")
        print("Nome: ", prod[1])
        print("Preço: ", prod[2])
        achou = True
        break
if not achou:
    print("Código não encontrado no catálogo.")
```

```python
estoque = []
for i in range(5):
    print("Item", i + 1)
    item = {}
    item["codigo"] = int(input("Código: "))
    item["nome"] = input("Nome: ")
    item["quantidade"] = int(input("Quantidade: "))
    item["preco"] = float(input("Preço: "))
    estoque.append(item)

print("(a) Itens Cadastrados")
for item in estoque:
    print(item)

valor_total_estoque = 0
maior_valor_item = 0
item_maior = None

for item in estoque:
    valor_item = item["quantidade"] * item["preco"]
    valor_total_estoque += valor_item

    if valor_item > maior_valor_item:
        maior_valor_item = valor_item
        item_maior = item

print("(b) Valor Total do Estoque")
print("Total: ", valor_total_estoque)

print("(c) Item de Maior Valor em Estoque")
if item_maior is not None:
    print("Item:", item_maior["nome"], "Valor em estoque: ", maior_valor_item)

resumo = (len(estoque), valor_total_estoque)
print("(d) Tupla Resumo")
print("Resumo:", resumo)
```
